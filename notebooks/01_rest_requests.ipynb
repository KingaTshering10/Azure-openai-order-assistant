# 1 · Call the local order API

Start `python GET_JSON.py` in a separate terminal. These REST requests need no Azure credentials. The API contains one mock order, `123456789`; its 2024 dates are historical sample data.

```python
from pathlib import Path
import sys
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "assistant_core.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository.")
sys.path.insert(0, str(ROOT))
```

```python
import os
import requests
from dotenv import load_dotenv
load_dotenv(ROOT / ".env")
api_url = os.getenv("LOCAL_API_URL") or "http://127.0.0.1:8000/test"
response = requests.get(api_url, params={"order_number": "123456789"}, timeout=15)
response.raise_for_status()
data = response.json()
data
```

```python
print("Expected delivery:", data["EstimatedDeliveryDate"])
print("Order total:", data["OrderTotal"], data["Currency"])
print("Shipping share of total:", data["ShippingPercentageOfTotal"], "%")
```

## Unknown order
A lookup for a different order returns HTTP 404 rather than silently returning the sample order.

```python
missing = requests.get(api_url, params={"order_number": "999999999"}, timeout=15)
print("HTTP status:", missing.status_code)
print(missing.json())
```

For the interactive endpoint schema, open http://127.0.0.1:8000/docs while the server is running. A request without an order number returns the sample record for compatibility with the original GET exercise.
